```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import os
import matplotlib.cm as cm
from matplotlib import gridspec
import matplotlib.pyplot as plt
plt.rcParams.update({'font.size': 22})
```

```python
path = "../Data/Simulations/Bridge 20 Structure Factor Cluster Size/"
filenames = os.listdir(path)
std_lst = []
size_lst = []
for i in range(len(filenames)):
    df = pd.read_csv(path + filenames[i], sep=None, engine="python", comment="#", skip_blank_lines=True, skiprows=0, header=None)
    df.columns = ["Index", "Cluster Identifier" ,"Cluster Size", "Radius of Gyration" ,"Gyration Tensor.XX", "Gyration Tensor.YY", "Gyration Tensor.ZZ", "Gyration Tensor.XY", "Gyration Tensor.XZ" ,"Gyration Tensor.YZ"]
    mean_cluster = np.mean(df["Radius of Gyration"])
    std_cluster = np.std(df["Radius of Gyration"])
    size_lst.append(mean_cluster)
    std_lst.append(std_cluster)
size = np.array(size_lst)
std = np.array(std_lst)
print('size: ', size[-1]*26*2*np.sqrt(5/2))
print('std: ', std[-1]*26*2)
```

```python
x = np.linspace(1, 200, len(filenames))
fig, ax = plt.subplots(figsize=(10, 7))
#plt.plot(x, size*26*2*np.sqrt(5/2), linewidth=3)
plt.errorbar(x, size*26*2*np.sqrt(5/2), yerr=std*26*2)
plt.xlabel('Simulation Frame')
plt.ylabel('Mean Cluster Diameter (nm)')
```

## Make Crystal Movie

```python
path = "../Data/Simulations/Bridge 20 Crystalinity Analysis/"
filenames = (os.listdir(path))
for i in range(len(filenames)):
    if len(filenames[i]) == 14:
        new_name = filenames[i][0:13] + '00' + filenames[i][-1]
        os.rename(path + filenames[i], path + '/' + new_name)
    elif len(filenames[i]) == 15:
        new_name = filenames[i][0:13] + '0' + filenames[i][-2:]
        os.rename(path + filenames[i], path + '/' + new_name)
```

```python
path = "../Data/Simulations/Bridge 20 Crystalinity Analysis/"
filenames = sorted(os.listdir(path))
for i in range(len(filenames)):
    df = pd.read_csv(path + filenames[i], sep=None, engine="python", comment="#", skip_blank_lines=True, skiprows=0, header=None).values[:,1]
    percent_crystal_row = df/np.sum(df)
    if i == 0:
        percent_crystal = percent_crystal_row.reshape(1,-1)
    else:
        percent_crystal = np.vstack((percent_crystal, percent_crystal_row.reshape(1,-1)))
percent_crystal = percent_crystal*100
```

```python
frame = np.linspace(1, len(filenames)*2, len(filenames))
#for i in range(len(filenames)):
i = 1 
fig, ax = plt.subplots(figsize=(10, 7))
plt.plot(frame[:i], percent_crystal[0:i,0], linewidth=5, label = 'Amorphous', color = "#636363")
plt.scatter(frame[:i], percent_crystal[0:i,0], linewidth=3, color = "#636363")
plt.plot(frame[:i], percent_crystal[0:i,1], linewidth=5, label = 'FCC', color = "#66ff66")
plt.scatter(frame[:i], percent_crystal[0:i,1], linewidth=3, color = "#66ff66")
plt.plot(frame[:i], percent_crystal[0:i,2], linewidth=5, label = 'HCP', color = "#ff6666")
plt.scatter(frame[:i], percent_crystal[0:i,2], linewidth=3, color = "#ff6666")
plt.xlabel('Simulation Frame')
plt.ylabel('Percent of Particles')
plt.xlim([0, 200])
plt.legend(loc = 'upper right')
plt.savefig('../Figures/Crystal 20 Simulation/plot_' + str(i) + '.png', dpi=600, bbox_inches="tight")
plt.close()
```
